```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 2 * np.pi, 100)

a = 5
b = 1.5

def f(x, a, b):
    return a * np.exp(-b * x)
```

```python
y = f(x, a, b)

plt.plot(x, y)

plt.xlabel("x")
plt.ylabel("y = a * exp(-b * x)")
plt.title("Function")

plt.savefig("plot.pdf")

plt.show()
```
