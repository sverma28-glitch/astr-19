```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0.1, 2 * np.pi, 100)

a = 5
b = 1.5

def f(x, a, b):
    return a * np.exp(-b * x)
```

```python
y = f(x, a, b)

plt.plot(x, y)

plt.xlabel("x")
plt.ylabel("y = a * exp(-b * x)")
plt.title("Function")

plt.savefig("plot.pdf")

plt.show()
```

```python
sigma = np.linspace(2, 0.2, len(x))

scatter = np.random.normal(0, sigma)

y_scattered = y + scatter

plt.errorbar(x, y_scattered, yerr=sigma, fmt='o', color='pink')

plt.xlabel("x")
plt.ylabel("y")

plt.savefig("scattered.pdf")
plt.show()
```

```python
from scipy.optimize import curve_fit

def exp(x, a, b):
    return a * np.exp(-b * x)

def power(x, a, b):
    return a / (x ** b)

exp_params, exp_cov = curve_fit(exp, x, y_scattered)
power_params, power_cov = curve_fit(power, x, y_scattered)

print(f"Exponential model: a = {exp_params[0]:.3f}, b = {exp_params[1]:.3f}")
print(f"Power model: a = {power_params[0]:.3f}, b = {power_params[1]:.3f}")

plt.errorbar(x, y_scattered, yerr=sigma, fmt='o', color='pink', label='Data')

plt.plot(x, exp(x, exp_params[0], exp_params[1]), label='Exponential fit')
plt.plot(x, power(x, power_params[0], power_params[1]), label='Power fit')

plt.xlabel("x")
plt.ylabel("y")
plt.title("Best Fit Models")
plt.legend()

plt.savefig("bestfitmodels.pdf")
plt.show()
```
